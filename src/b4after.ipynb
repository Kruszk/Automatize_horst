```python
def plot_paired_spectrums_per_file(input_folder: str):
    """
    Processes each ROOT file and creates a two-subplot figure per file:
    Left = Original Spectrum, Right = Reconstructed Spectrum.
    """
    folder = Path(input_folder)
    root_files = list(folder.rglob("*.root"))

    if not root_files:
        print(f"Error: No .root files found in '{input_folder}'.")
        return

    print(f"Found {len(root_files)} ROOT file(s). Generating paired plots...\n")

    for file_path in tqdm(root_files, desc="Processing Files", unit="file"):
        try:
            with uproot.open(file_path) as f:
                has_spec = "spectrum" in f
                has_rec = "spectrum_reconstructed" in f

                if not has_spec and not has_rec:
                    tqdm.write(f"Skipping {file_path.name}: Spectra not found.")
                    continue

                # Create 2 subplots side-by-side for this file
                fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)
                fig.suptitle(f"File: {file_path.name}", fontsize=12, fontweight="bold")

                # --- Plot 1: Spectrum ---
                if has_spec:
                    h1 = f["spectrum"]
                    vals1 = h1.values()
                    errs1 = h1.errors()
                    edges1 = h1.axis().edges()
                    centres1 = (edges1[:-1] + edges1[1:]) / 2

                    ax1.stairs(vals1, edges1, color="royalblue", label="Original")
                    ax1.errorbar(centres1, vals1, yerr=errs1, fmt=".", color="royalblue", alpha=0.5, capsize=0)
                    ax1.set_title("spectrum")
                    ax1.set_xlabel("Channels / Energy")
                    ax1.set_ylabel("Counts")
                    ax1.grid(True, linestyle="--", alpha=0.5)
                else:
                    ax1.text(0.5, 0.5, "Not Found", ha="center", va="center")

                # --- Plot 2: Spectrum Reconstructed ---
                if has_rec:
                    h2 = f["spectrum_reconstructed"]
                    vals2 = h2.values()
                    errs2 = h2.errors()
                    edges2 = h2.axis().edges()
                    centres2 = (edges2[:-1] + edges2[1:]) / 2

                    ax2.stairs(vals2, edges2, color="crimson", label="Reconstructed")
                    ax2.errorbar(centres2, vals2, yerr=errs2, fmt=".", color="crimson", alpha=0.5, capsize=0)
                    ax2.set_title("spectrum reconstructed")
                    ax2.set_xlabel("Channels / Energy")
                    ax2.grid(True, linestyle="--", alpha=0.5)
                else:
                    ax2.text(0.5, 0.5, "Not Found", ha="center", va="center")

                plt.tight_layout()
                plt.show()

        except Exception as err:
            tqdm.write(f"Error reading {file_path.name}: {err}")


if __name__ == "__main__":
    FOLDER_PATH = "/home/kacper-ruszkowski/HIGS/Higs2026/workplace/WORKDIR/Horst_output"
    
    plot_paired_spectrums_per_file(FOLDER_PATH)
```

```python
import re
from pathlib import Path
import uproot
import matplotlib.pyplot as plt

# Configuration
FOLDER_PATH = Path("/home/kacper-ruszkowski/HIGS/Higs2026/workplace/WORKDIR/Horst_output")
OBJECT_PATH = "topdown/topdown_FEP"

# Regex pattern matching energy, date (8 digits), width, and offset
filename_regex = re.compile(
    r"^(?P<energy>\d+(?:\.\d+)?(?:MeV|keV))_.*?(?P<date>\d{8})_.*_(?P<width>\d+mm)_(?P<offset>-?\d+mm)\.root$"
)

# Helper function to convert energy string to keV for proper numerical sorting
def energy_in_kev(energy_str: str) -> float:
    val = float(re.search(r"[\d.]+", energy_str).group())
    return val * 1000 if "MeV" in energy_str else val

data_by_group = {}

# 1. Scan directory and extract object data
for file_path in FOLDER_PATH.glob("*.root"):
    match = filename_regex.match(file_path.name)
    if not match:
        continue

    energy = match.group("energy")
    date = match.group("date")
    width = match.group("width")
    offset = match.group("offset")
    
    param_key = f"{width}_{offset}"
    group_key = (energy, date)

    try:
        with uproot.open(file_path) as f:
            if OBJECT_PATH not in f:
                print(f"Warning: Object '{OBJECT_PATH}' not found in {file_path.name}")
                continue

            obj = f[OBJECT_PATH]

            if hasattr(obj, "to_numpy"):  # TH1 Histogram
                values, edges = obj.to_numpy()
                x = (edges[:-1] + edges[1:]) / 2
                y = values
            else:  # TGraph / TGraphErrors
                x = obj.member("fX")
                y = obj.member("fY")

            if group_key not in data_by_group:
                data_by_group[group_key] = {}

            data_by_group[group_key][param_key] = (x, y)

    except Exception as err:
        print(f"Error reading {file_path.name}: {err}")


# 2. Render each (energy, date) group in its own separate figure
sorted_groups = sorted(
    data_by_group.items(), 
    key=lambda item: (energy_in_kev(item[0][0]), item[0][1])
)

if not sorted_groups:
    print("No matching ROOT files found.")
else:
    for (energy, date), params in sorted_groups:
        plt.figure(figsize=(9, 5))

        for param_key, (x, y) in sorted(params.items()):
            plt.plot(x, y, label=param_key, linewidth=1.5)

        # Format 20260731 -> 2026-07-31 for title readability
        formatted_date = f"{date[:4]}-{date[4:6]}-{date[6:]}" if len(date) == 8 else date
        
        plt.title(f"Topdown FEP - {energy} ({formatted_date})", fontsize=13, fontweight="bold")
        plt.xlabel("Position / Bin Center", fontsize=11)
        plt.ylabel("Counts / Intensity", fontsize=11)
        plt.grid(True, linestyle="--", alpha=0.6)
        plt.legend(title="Width_Offset", loc="best", fontsize=9)
        plt.tight_layout()
        plt.show()
```
